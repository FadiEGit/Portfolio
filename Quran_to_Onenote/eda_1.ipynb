```python
import requests
import json
```

# English Translation

```python
for i in range(1, 115):
    url = f'https://raw.githubusercontent.com/semarketir/quranjson/master/source/translation/en/en_translation_{i}.json'
    response = requests.get(url)
    
    if response.status_code == 200:
        with open(f'en_translation/en_translation_{i}.json', 'w', encoding='utf-8') as file:
            file.write(response.text)
        print(f'Downloaded: en_translation_{i}.json')
    else:
        print(f'Failed to download: {url}')
```

```python
import os

# Initialize a global variable to hold the full text
full_text = ""

# Iterate through the files in numerical order from 1 to 114
for i in range(1, 115):
    file_name = f'en_translation_{i}.json'
    file_path = os.path.join('en_translation', file_name)
    with open(file_path, 'r', encoding='utf-8') as f:
        json_data = json.load(f)
        verses = json_data.get('verse', {})
        for verse_key, verse_text in verses.items():
            full_text += verse_text + " "
```

```python
with open('full_text.txt', 'w', encoding='utf-8') as f:
    f.write(full_text)
```

```python
from collections import Counter
import re

# Split the text into words using regex to handle punctuation and case
words = re.findall(r'\b\w+\b', full_text.lower())

# Create a word count dictionary
word_count = Counter(words)


with open('word_count.txt', 'w', encoding='utf-8') as f:
    for word, count in word_count.items():
        f.write(f'{word}: {count}\n')
```

## Downloading full Arabic Quran

```python
for i in range(1, 115):
    url = f'https://raw.githubusercontent.com/semarketir/quranjson/master/source/surah/surah_{i}.json'
    response = requests.get(url)
    
    if response.status_code == 200:
        with open(f'data/surah/surah_{i}.json', 'w', encoding='utf-8') as file:
            file.write(response.text)
        print(f'Downloaded: surah_{i}.json')
    else:
        print(f'Failed to download: {url}')
```

```python
import os

# Initialize a global variable to hold the full text
full_text = ""

# Iterate through the files in numerical order from 1 to 114
for i in range(1, 115):
    file_path = f'data/surah/surah_{i}.json'
    with open(file_path, 'r', encoding='utf-8') as f:
        json_data = json.load(f)
        verses = json_data.get('verse', {})
        for verse_key, verse_text in verses.items():
            full_text += verse_text + " "
```

```python
with open('data/full_text.txt', 'w', encoding='utf-8') as f:
    f.write(full_text)
```

```python
from collections import Counter
import re

# Read the Arabic text from data/full_text.txt
with open('data/full_text.txt', 'r', encoding='utf-8') as f:
    arabic_text = f.read()

# Split the text into words using regex to handle Arabic characters
arabic_words = re.findall(r'\b[\u0600-\u06FF]+\b', arabic_text)

# Create a word count dictionary
arabic_word_count = Counter(arabic_words)

# Write the word count to data/word_count.txt
with open('data/word_count.txt', 'w', encoding='utf-8') as f:
    for word, count in arabic_word_count.items():
        f.write(f'{word}: {count}\n')
```

```python
# Sort the word count dictionary by count in descending order
sorted_arabic_word_count = sorted(arabic_word_count.items(), key=lambda x: x[1], reverse=True)

# Write the sorted word count to data/word_count.txt
with open('data/word_count.txt', 'w', encoding='utf-8') as f:
    for word, count in sorted_arabic_word_count:
        f.write(f'{word}: {count}\n')
```
