```python
# plot every column in weather_df and save to file
import matplotlib.pyplot as plt
import seaborn as sns
import os
import numpy as np

def plot_weather(df, column):   
    plt.figure(figsize=(10, 6))
    sns.histplot(df[column], bins=30, kde=True)
    plt.title(f'Distribution of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')
    plt.grid()
    plt.savefig(f'weather_columns/weather_{column}.png')
    plt.close()
```

```python
# Create a directory to save the plots
if not os.path.exists('weather_columns'):
    os.makedirs('weather_columns')
for column in weather_df.columns:
    plot_weather(weather_df, column)
```
